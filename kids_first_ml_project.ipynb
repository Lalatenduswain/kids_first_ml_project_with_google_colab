## Step 2 — 🥭🍊 ଫଳ ଡାଟା ତିଆରି (Collect Evidence!)

> 🕵️ **Detective Robot:** "ଫଳ ଚିହ୍ନିବା ପାଇଁ ଆଗେ ଦେଖ — ଏହା ଆମ୍ବ ନା କମଳା?"
>
> **Clue 1 🔍:** ଆମ୍ବ ହାଲୁକା (80–110g), କମଳା ଭାରୀ (130–200g)!
> 
> **Clue 2 🎨:** ଆମ୍ବ ହଳଦିଆ (color=0), କମଳା... କମଳା ରଙ୍ଗ (color=1)!

| ଫଳ | ଓଜନ (ଗ୍ରାମ) | ରଙ୍ଗ code | ସନ୍ଦେହ ?
|---|---|---|---|
| 🍊 କମଳା | ଭାରୀ (130–200g) | 1 | "ମୁଁ ଗୋଲ୍ ଆଉ ଭାରୀ! 😤" |
| 🥭 ଆମ୍ବ | ହାଲୁକା (80–110g) | 0 | "ମୁଁ slim ଆଉ ହଳଦିଆ! 😎" |

````python
# 🗂️ CASE FILE — Fruit Evidence Database
# ========================================
# format: [ଓଜନ(gram), ରଙ୍ଗ(0=ହଳଦିଆ, 1=କମଳା)]
#
# 🤖 Robot whispers: "ଏଇଟି ଆମ AI ର question paper ଭଳି!"

fruits_data = [
    # ଓଜନ, ରଙ୍ଗ
    [150,    1],   # 🍊 କମଳା 1  — "ମୁଁ ବହୁତ ଭାରୀ ଆଉ ପୂରା ଗୋଲ!"
    [170,    1],   # 🍊 କମଳା 2  — "ମୁଁ ତା'ଠାରୁ ଆହୁରି ଭାରୀ! 💪"
    [140,    1],   # 🍊 କମଳା 3  — "ମୁଁ ଆଜି ଟିକେ diet ରେ ଅଛି"
    [160,    1],   # 🍊 କମଳା 4  — "ମୁଁ ପୂରା average!"
    [180,    1],   # 🍊 କମଳା 5  — "ମୁଁ Gym ଯାଏ 💪💪"
    [200,    1],   # 🍊 କମଳା 6  — "BOSS of oranges! 👑"
    [130,    1],   # 🍊 କମଳା 7  — "Smallest orange, still proud"
    [ 90,    0],   # 🥭 ଆମ୍ବ 1  — "ମୁଁ ଫଳର ରାଜା ଆମ୍ବ! 👑"
    [100,    0],   # 🥭 ଆମ୍ବ 2  — "ମୁଁ ଖାସ୍ Alphonso ବଂଶର 😌"
    [ 80,    0],   # 🥭 ଆମ୍ବ 3  — "Lightweight champion! 🥊"
    [110,    0],   # 🥭 ଆମ୍ବ 4  — "ମୁଁ ସବୁଠାରୁ ଓଜନିଆ ଆମ୍ବ 😤"
    [ 95,    0],   # 🥭 ଆମ୍ବ 5  — "ମୁଁ ଠିକ୍ ଠାକ୍ ଅଛି"
    [105,    0],   # 🥭 ଆମ୍ବ 6  — "Round ଆଉ yummy! 😋"
    [ 85,    0],   # 🥭 ଆମ୍ବ 7  — "Smallest mango, big taste!"
]

# ✅ ଉତ୍ତର list — 0=ଆମ୍ବ, 1=କମଳା
labels = [1, 1, 1, 1, 1, 1, 1,   # ← ସବୁ 7 ଟି କମଳା (Team Orange 🍊)
          0, 0, 0, 0, 0, 0, 0]   # ← ସବୁ 7 ଟି ଆମ୍ବ  (Team Mango 🥭)

X = np.array(fruits_data)
y = np.array(labels)

print("🗂️  EVIDENCE COLLECTED!")
print("=" * 38)
print(f"🥭 Team Mango  (ଆମ୍ବ team) : {sum(y==0)} ଟି")
print(f"🍊 Team Orange (କମଳା team): {sum(y==1)} ଟି")
print(f"📦 ମୋଟ suspects           : {len(y)} ଟି")
print("=" * 38)
print("🕵️  Detective Robot: 'ଚାଲ ଟୁକୁ! ମିଶିକି Case solve କରିବା!'")
````

## Step 3 — 📊 Suspect Chart (Suspect Lineup!)

> 🕵️ **Detective Robot:** "ସବୁ ଫଳଗୁଡ଼ିକୁ ଧାଡ଼ିରେ ଠିଆ କର! ଆମେ ଦେଖିବା — ଆମ୍ବ ହାଲୁକା, କମଳା ଭାରୀ!"
>
> *(ଯଦି ଫଳ chart ଦେଖି ଠିକ୍‌ରେ ଚିହ୍ନି ହେବ, ତେବେ AI ସଫଳ! 🎯)*

````python
# 🎭 SUSPECT LINEUP CHART!
# ========================

plt.figure(figsize=(9, 5))

mango_weights  = [X[i][0] for i in range(len(y)) if y[i] == 0]
orange_weights = [X[i][0] for i in range(len(y)) if y[i] == 1]

# ଆମ୍ବ — ହଳଦିଆ dots
plt.scatter(range(len(mango_weights)), mango_weights,
            color='gold', s=300, label='🥭 ଆମ୍ବ (Slim guys!)', zorder=3,
            edgecolors='darkorange', linewidths=2)

# କମଳା — ଲାଲ dots
plt.scatter(range(len(orange_weights)), orange_weights,
            color='darkorange', s=300, label='🍊 କମଳା (Heavy squad!)', zorder=3,
            edgecolors='red', linewidths=2)

# ଦୁଇ team ମଝିରେ ଥିବା line
plt.axhline(y=120, color='red', linestyle='--', alpha=0.5, linewidth=2,
            label='🚔 Boundary (120g)')

plt.title('🥭 ଆମ୍ବ vs 🍊 କମଳା — Weight Showdown! 💪', fontsize=15, pad=12)
plt.ylabel('ଓଜନ (ଗ୍ରାମ) ↑ ଭାରୀ',  fontsize=12)
plt.xlabel('ଫଳ ସଂଖ୍ୟା (Fruit No.)', fontsize=12)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("💡 ଦେଖିଲୁ ଟୁକୁ! ଦୁଇଟି ଯାକ team ପୂରା ଅଲଗା ଦିଶୁଛନ୍ତି!")
print("🤖 Robot: 'ଏହା ଦେଖି ମୋ brain ପୂରା ଖୁସି ହୋଇଗଲା! 🥳'")
````

## Step 4 — 🧠 AI Brain ON! (Training Time!)

> 🤖 **Robot:** "ଏବେ ମୁଁ ଶିଖିବି! ଠିକ୍ ଯେମିତି:"
>
> - 🍜 ତୁ ଖାଇ ଖାଇ ଶିଖୁ (experience)
> - 📚 ମୁଁ data ଦେଖି ଶିଖେ (training!)
>
> **🎓 School:** Data feeding into AI brain = 0.001 seconds!
> *(ତୁ ୧୦ ବର୍ଷ ଧରି ଶିଖୁଛୁ... ଆଉ ମୁଁ ଗୋଟିଏ flash ରେ! 😏)*
>
> *(ହଁ ହଁ, exam ରେ cheat କରନା। AI ଶିଖ!)*

````python
# 🧠 BRAIN ACTIVATION SEQUENCE
# =============================
import time

print("🔌 AI ର brain ଏବେ start ହେଉଛି...")
print("⚡ Loading... [██████████] 100% — DONE! (actually instant 😅)")
print()

# Decision Tree — ଜଣେ ଭଲ detective ଭଳି ଶାଖା ପ୍ରଶାଖା କରି ଚିନ୍ତା କରେ! 🌳
model = DecisionTreeClassifier(random_state=42)

# 🎓 TRAINING — AI ଏବେ data ଦେଖି ନୂଆ କଥା ଶିଖୁଛି!
model.fit(X, y)

# AI ର ନିଜର self-test
predictions = model.predict(X)
accuracy    = accuracy_score(y, predictions) * 100

print("🎓 TRAINING REPORT CARD")
print("=" * 35)
print(f"📝 AI ର Marks   : {accuracy:.0f}% 📊")
print(f"📚 Questions    : {len(y)} ଟି ଫଳ")
print(f"✅ Correct      : {int(accuracy/100*len(y))} ଟି")
print("=" * 35)

if accuracy == 100:
    print()
    print("🎉 ୧୦୦%!! ଆମ AI ପୂରା TOPPER ହୋଇଗଲା!!")
    print("🤖 Robot: 'ସାର୍ ମୋତେ Gold Medal ଦିଅନ୍ତୁ! 🥇'")
    print("🏫 School: 'ତୁ ତ ଗୋଟେ Robot... ତୋତେ Medal ମିଳିବନି 😤'")
    print("🤖 Robot: '...okay fine 😞'")
else:
    print(f"\n📖 AI ଆହୁରି ଶିଖୁଛି... {accuracy:.0f}% ହୋଇଗଲା!")
    print("🤖 Robot: 'ମୋତେ ଆହୁରି data ଦିଅ! ଆହୁରି ଦିଅ!'")
````

````python
# 🔍 MYSTERY FRUIT INVESTIGATION!
# ================================
# ✏️ ଏଠି ନିଜ ମନ ପସନ୍ଦର ଫଳ ଭରି ଦିଅ!
#    ଓଜନ: gram ରେ ଲେଖ
#    ରଙ୍ଗ: 0=ହଳଦିଆ (ଆମ୍ବ ଭଳି), 1=କମଳା ରଙ୍ଗ

new_fruits = [
    [155, 1],   # 🎭 Mystery Fruit A — "ମୁଁ ବହୁତ ଭାରୀ ଆଉ ଲାଲ ରଙ୍ଗର!"
    [ 92, 0],   # 🎭 Mystery Fruit B — "ମୁଁ ହାଲୁକା ଆଉ ହଳଦିଆ ରଙ୍ଗର!"
    [175, 1],   # 🎭 Mystery Fruit C — "ମୁଁ gym ଯାଏ! ମୋର ଅନେକ ବଳ!"
    [ 88, 0],   # 🎭 Mystery Fruit D — "ଚୁପ୍ ଚୁପ୍... (hiding)"
]

fruit_names  = ['A', 'B', 'C', 'D']
fruit_emojis = ['🎭', '🥸', '🤔', '😶']
label_map    = {0: '🥭 ଆମ୍ବ (MANGO!)',
                1: '🍊 କମଳା (ORANGE!)'}

results = model.predict(np.array(new_fruits))

print("🔍 DETECTIVE REPORT — MYSTERY FRUITS EXPOSED! 🎉")
print("=" * 48)
print(f"{'Fruit':<8} {'ଓଜନ':>6} {'ରଙ୍ଗ':>5}  →  ପରିଚୟ")
print("-" * 48)

for i, (fruit, result) in enumerate(zip(new_fruits, results)):
    color_name = 'ହଳଦିଆ' if fruit[1] == 0 else 'କମଳା'
    print(f"{fruit_emojis[i]} Fruit {fruit_names[i]}:  {fruit[0]}g  {color_name:>6}  →  {label_map[result]}")

print("=" * 48)
print()
print("🤖 Robot: 'ମୁଁ ସବୁ ଚିହ୍ନଟ କରିଦେଲି! ଏବେ ମୋତେ ଆଇସ୍କ୍ରିମ୍ ମିଳିବ ତ? 🍦'")
print("👨‍🔬 Scientist: 'ଆରେ ତୁ ତ ଗୋଟେ ଯନ୍ତ୍ର... ତୋତେ କେମିତି ଆଇସ୍କ୍ରିମ୍ ଖୁଆଇବି! 😂'")
````

## Step 7 — 🔍 RAG: Robot ର "Open Book" Trick!

> 🤖 **Robot:** "ମୋ ମୁଣ୍ଡରେ ସବୁକିଛି ମନେ ନଥାଏ... କିନ୍ତୁ ମୋ **book** ରେ ସବୁ ଅଛି! ପ୍ରଶ୍ନ ଆସିଲେ ମୁଁ ଠିକ୍ page ଖୋଜେ, ତା'ପରେ ପଢ଼ି ଉତ୍ତର ଦିଏ! 📖✨"
>
> **RAG = Retrieval (ଖୋଜିବା) + Generation (ଉତ୍ତର ଦେବା)**
> *(ଠିକ୍ ଯେମିତି open-book exam — ମନେ ରଖିବା ଦରକାର ନାହିଁ, ବାସ୍ ଖୋଜିବା ଜାଣିଲେ ହେଲା! 😎)*

````python
# 🔍📖 RAG — Robot ର "Open Book" Superpower!
# ============================================
# (ନୂଆ install ନାହିଁ — ଆମ ପୁରୁଣା sklearn ହିଁ ବ୍ୟବହାର କରୁଛୁ!)

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# 📖 STEP 1 — ROBOT ର BOOK (knowledge base = ଫଳ facts ର page)
fruit_book = [
    "ଆମ୍ବ ହାଲୁକା ହୁଏ, ଓଜନ 80 ରୁ 110 gram. Mango is light, 80 to 110 grams.",
    "କମଳା ଭାରୀ ହୁଏ, ଓଜନ 130 ରୁ 200 gram. Orange is heavy, 130 to 200 grams.",
    "ଆମ୍ବ ର ରଙ୍ଗ ହଳଦିଆ, color code 0. Mango colour is yellow, code 0.",
    "କମଳା ର ରଙ୍ଗ କମଳା, color code 1. Orange colour is orange, code 1.",
    "120 gram ହେଉଛି boundary line: ତା ଉପରେ କମଳା, ତଳେ ଆମ୍ବ. 120g is the boundary.",
    "Alphonso ହେଉଛି ଆମ୍ବ ର ରାଜା ବଂଶ. Alphonso is the king variety of mango.",
]

# 🔢 STEP 2 — book ର ପ୍ରତ୍ୟେକ line କୁ "meaning fingerprint" (numbers) ରେ ବଦଳାଅ
vectorizer   = TfidfVectorizer()
book_vectors = vectorizer.fit_transform(fruit_book)
print("📚 Robot ର book ready! ", len(fruit_book), "ଟି page ମନେ ଅଛି.\n")

# 🔍 STEP 3+4 — RETRIEVAL (ଠିକ page ଖୋଜ) + GENERATION (ଉତ୍ତର ତିଆର)
def robot_answer(question):
    q_vec  = vectorizer.transform([question])          # ପ୍ରଶ୍ନ ର fingerprint
    scores = cosine_similarity(q_vec, book_vectors)[0] # ପ୍ରତ୍ୟେକ page ସହ ମେଳ କେତେ?
    best   = scores.argmax()                           # ସବୁଠୁ ଭଲ page
    fact   = fruit_book[best]

    print(f"❓ ପ୍ରଶ୍ନ  : {question}")
    print(f"🔍 Robot book ଖୋଜିଲା (match score = {scores[best]:.2f})")
    print(f"   📖 Page: {fact}")
    print(f"🤖 Robot : '{fact.split('.')[0].strip()} — ଏଇଟି ମୋ ଉତ୍ତର!'")
    print("-" * 58)

# 🎤 ROBOT କୁ ପ୍ରଶ୍ନ ପଚାର! (ନିଜ ପ୍ରଶ୍ନ ମଧ୍ୟ ଲେଖି ଦେଖ ✏️)
robot_answer("କମଳା ର ଓଜନ କେତେ?")        # How heavy is an orange?
robot_answer("ଆମ୍ବ ର ଓଜନ କେତେ?")        # How heavy is a mango?
robot_answer("ଆମ୍ବ ର ରଙ୍ଗ କ'ଣ?")          # What colour is a mango?
robot_answer("କମଳା ର color code କେତେ?")  # What is the orange's colour code?
robot_answer("Alphonso କ'ଣ?")             # What is Alphonso?
robot_answer("boundary line କେଉଁଠି?")     # Where is the boundary line?

print("💡 ଦେଖିଲୁ? Robot କିଛି ମନେ ରଖି ନାହିଁ — ପ୍ରତ୍ୟେକ ଥର book ଖୋଜି ଉତ୍ତର ଦେଲା.")
print("   ➜ ଏଇଟି ହିଁ RAG! book ବଦଳାଇ ଦେ, ଉତ୍ତର ନିଜେ ବଦଳିଯିବ. କୌଣସି re-training ନାହିଁ! 🎉")
````

## Step 8 — 🔗 PCP: Robot ର "Step-by-Step" ଚିନ୍ତା!

> 🤖 **Robot:** "ବଡ଼ ପ୍ରଶ୍ନକୁ ଏକାଥରେ solve କରିବା କଷ୍ଟ! ତେଣୁ ମୁଁ ତାକୁ **ଛୋଟ ଛୋଟ step** ରେ ଭାଙ୍ଗିଦିଏ। ପ୍ରତି step ର ଉତ୍ତରକୁ ମୁଁ ପରବର୍ତ୍ତୀ step କୁ ଦିଏ — ପୂରା relay race ଭଳି! 🏃‍♂️→🏃‍♀️→🏃"
>
> **PCP = Prompt Chaining Pattern** (ଗୋଟିଏ chain ରେ ଛୋଟ ଛୋଟ step)
>
> **ଆମ chain:** 🥎 ମାପ → 🥎 ତୁଳନା → 🥎 ନିଷ୍ପତ୍ତି → 🥎 ବୁଝାଅ
>
> *(ଗୋଟିଏ step ର baton ଅନ୍ୟ step କୁ pass ହୁଏ — ଯଦି କୌଣସି step ରେ baton ଖସିପଡ଼େ, ତେବେ ଆମେ ସହଜରେ ଧରିପାରିବା ଯେ କେଉଁଠି ଭୁଲ୍ ହେଲା! 🔍)*

````python
# 🔗🧠 PCP — Robot ର "Step-by-Step" ଚିନ୍ତା କରିବାର ଉପାୟ (Prompt Chaining Pattern)
# ==================================================================
# ଗୋଟିଏ ବଡ଼ କାମକୁ ଆମେ 4 ଟି ଛୋଟ step ରେ ଭାଙ୍ଗିଦେଲୁ।
# ପ୍ରତ୍ୟେକ step ର output ➜ ପରବର୍ତ୍ତୀ step ର input (baton pass କଲା ଭଳି 🥎)

def step1_measure(fruit):
    weight, color = fruit
    print(f"🥎 Step 1 — ମାପ    : ଓଜନ = {weight}g, ରଙ୍ଗ code = {color}")
    return weight, color                      # baton ➜ step 2

def step2_compare(weight, color):
    heavy = weight >= 120
    print(f"🥎 Step 2 — ତୁଳନା  : 120g boundary ସହ → {'ଭାରୀ 💪' if heavy else 'ହାଲୁକା 🪶'}")
    return heavy, color                       # baton ➜ step 3

def step3_decide(heavy, color):
    guess = "🍊 କମଳା" if (heavy and color == 1) else "🥭 ଆମ୍ବ"
    print(f"🥎 Step 3 — ନିଷ୍ପତ୍ତି: ଏଇଟି {guess}")
    return guess                              # baton ➜ step 4

def step4_explain(guess):
    reason = "ଭାରୀ ଆଉ କମଳା ରଙ୍ଗର" if guess == "🍊 କମଳା" else "ହାଲୁକା ଆଉ ହଳଦିଆ ରଙ୍ଗର"
    print(f"🥎 Step 4 — ବୁଝାଅ   : '{guess}, କାରଣ ଏହା {reason}!'")
    print("=" * 56)

# 🔗 CHAIN ଚଲାଅ — baton ଗୋଟିଏ ପରେ ଗୋଟିଏ pass ହେଉଛି!
def robot_thinks(fruit):
    print(f"🎭 Mystery Fruit ଆସିଗଲା: {fruit}")
    w, c  = step1_measure(fruit)   # 🥎 baton 1
    h, c  = step2_compare(w, c)    # 🥎 baton 2
    guess = step3_decide(h, c)     # 🥎 baton 3
    step4_explain(guess)           # 🥎 baton 4
    return guess

# 🕵️ ଏବେ ଦୁଇଟି mystery fruit ଉପରେ chain ଚଲାଇ ଦେଖିବା!
robot_thinks([175, 1])   # ଭାରୀ + କମଳା ରଙ୍ଗର
robot_thinks([ 88, 0])   # ହାଲୁକା + ହଳଦିଆ ରଙ୍ଗର

print("💡 ଦେଖିଲୁ ଟୁକୁ? ଗୋଟିଏ ବଡ଼ ପ୍ରଶ୍ନ ('ଏଇଟି କେଉଁ ଫଳ?') 4 ଟି ଛୋଟ step ରେ କେମିତି ଭାଙ୍ଗିଗଲା।")
print("   ➜ ଏଇଟି ହିଁ ଆମ PCP! ପ୍ରତ୍ୟେକ step ପୂରା ଛୋଟ, ସହଜ ଆଉ check କରିବା ପାଇଁ ବହୁତ ସୁବିଧା। 🔗")
````

## 🔗🔍 ଏ ଦୁଇଟି ମିଶିଲେ କ'ଣ ହୁଏ? (The Pro Combo!)
